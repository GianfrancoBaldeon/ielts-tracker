```python
import pandas as pd
import numpy as np
import statsmodels.api as sm
```

```python
Boeing=pd.read_csv("BoeingStock.csv")
```

```python
CocaC=pd.read_csv("CocaColaStock.csv")
```

```python
GE=pd.read_csv("GEStock.csv")
```

```python
IBM=pd.read_csv("IBMStock.csv")
PG=pd.read_csv("ProcterGambleStock.csv")
```

```python
Boeing["Date"] = pd.to_datetime(Boeing["Date"], format="%m/%d/%y")
CocaC["Date"] = pd.to_datetime(CocaC["Date"], format="%m/%d/%y")
GE["Date"] = pd.to_datetime(GE["Date"], format="%m/%d/%y")
IBM["Date"] = pd.to_datetime(IBM["Date"], format="%m/%d/%y")
PG["Date"] = pd.to_datetime(PG["Date"], format="%m/%d/%y")
```

```python
# 1)	¿Cuántas observaciones hay en cada data set?
# Respuesta 480
print(len(Boeing))
print(len(CocaC))
print(len(GE))
print(len(IBM))
print(len(PG))
```
